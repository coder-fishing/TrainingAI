# Foundation models (mô hình nền tảng)
Các mô hình học máy quy mô lớn được tiền huấn luyện (pre-trained) trên lượng dữ liệu khổng lồ bằng phương pháp tự giám sát (self-supervision), đóng vai trò nền tảng tri thức tổng quát để tùy biến cho nhiều tác vụ khác nhau

| Traditional ML Model | Foundation Model |
|---|---|
| Thường phục vụ một task cụ thể | Có thể phục vụ rất nhiều task |
| Dataset thường chuyên biệt | Pre-train trên dữ liệu rất lớn, đa dạng |
| Ví dụ: model dự đoán spam | Ví dụ: GPT |
| Muốn task mới thường phải train model mới | Có thể dùng prompting/fine-tuning cho task mới |


# LLM (Mô hình ngôn ngữ lớn)
Mạng nơ-rôn sâu (thường dựa trên kiến trúc Transformer) có hàng tỷ tham số, được huấn luyện trên tập dữ liệu văn bản khổng lồ với nhiệm vụ cốt lõi là dự đoán từ/token tiếp theo

## Ứng dụng của LLM 
- Sáng tạo nội dung & Sinh văn bản mới (Content Creation & Text Generation): Viết thơ, tiểu thuyết, bài báo, sáng tạo kịch bản và tự động viết mã máy tính (computer code)
- Dịch thuật máy (Machine Translation) & Phân tích cảm xúc (Sentiment Analysis): Dịch giữa các ngôn ngữ và đánh giá sắc thái cảm xúc trong văn bản
- Tóm tắt văn bản (Text Summarization): Tóm tắt các đoạn văn hoặc tài liệu dài thành nội dung ngắn gọn
- Chatbot & Trợ lý ảo thông minh: Các mô hình như ChatGPT hay Google Gemini (Bard) giúp trả lời câu hỏi trực tiếp, đồng thời tích hợp để nâng cao năng lực cho các công cụ tìm kiếm truyền thống như Google Search hay Microsoft Bing
- Truy xuất tri thức chuyên ngành (Knowledge Retrieval): Trích xuất và tìm kiếm thông tin trong các lĩnh vực phức tạp như Y tế và Pháp luật (đọc lướt khối lượng lớn tài liệu, tóm tắt các điều khoản/văn bản dài và giải đáp thắc mắc kỹ thuật)
- Tự động hóa tác vụ văn bản: Tự động hóa gần như bất kỳ quy trình làm việc nào liên quan đến việc phân tích, xử lý và tạo ra ngôn ngữ tự nhiên

## Các Bước Xây Dựng LLM
- Giai đoạn 1 - Pretraining (Tiền huấn luyện): Mô hình được huấn luyện trên tập dữ liệu văn bản thô khổng lồ (hàng tỷ đến hàng nghìn tỷ từ) thông qua nhiệm vụ dự đoán từ tiếp theo (next-word prediction) để tạo ra một Foundation Model (mô hình nền tảng)

- Giai đoạn 2 - Finetuning (Tinh chỉnh): Sử dụng tập dữ liệu có gắn nhãn nhỏ hơn để tinh chỉnh mô hình nền tảng phục vụ các nhiệm vụ cụ thể (như phân loại, tóm tắt, làm trợ lý cá nhân)
. Các mô hình tinh chỉnh chuyên biệt cho ngành nghề (như BloombergGPT cho tài chính hoặc mô hình cho y tế) thường đạt hiệu năng vượt trội hơn các mô hình đa dụng trong miền kiến thức đó

## (Using LLMs for different tasks)
- Hầu hết các LLM hiện đại đều dựa trên kiến trúc Transformer (xuất bản năm 2017)
- Kiến trúc Transformer gốc gồm 2 phần: Encoder (để phân tích văn bản) và Decoder (để sinh văn bản)
. Tuy nhiên, các mô hình sinh văn bản như GPT đã tối giản hóa bằng cách chỉ sử dụng các khối Decoder (Decoder-only)

## Khai thác các tập dữ liệu lớn (Utilizing large datasets)
- Việc tiền huấn luyện yêu cầu lượng dữ liệu khổng lồ (ví dụ tập dữ liệu GPT-3 gồm hàng trăm tỷ token từ nhiều nguồn như web, sách, Wikipedia)
. Tập dữ liệu khổng lồ này giúp mô hình xuất hiện các năng lực bộc phát (emergent properties) như khả năng lập luận, dịch thuật hay tóm tắt dù chỉ học qua nhiệm vụ đoán từ

## Building a Large Language Model Lộ trình xây dựng LLM

- Tiền xử lý dữ liệu & Lập trình cơ chế Attention: Chuẩn bị dữ liệu đầu vào (tokenization) và tự viết cơ chế tự chú ý (Self-Attention)

- Lập trình kiến trúc & Pretrain mô hình GPT: Cài đặt hoàn chỉnh mô hình GPT và chạy huấn luyện tiền xử lý trên tập dữ liệu nhỏ (phù hợp với máy tính cá nhân), hoặc nạp trọng số đã huấn luyện sẵn

- Finetuning thành trợ lý cá nhân / bộ phân loại: Tinh chỉnh mô hình đã pretrain để biết tuân thủ hướng dẫn người dùng (instruction-following) hoặc phân loại văn bản
.

# NLP (Natural Language Processing):
Lĩnh vực xử lý ngôn ngữ tự nhiên, đã chuyển dịch từ các mô hình quy tắc/thống kê truyền thống sang kỷ nguyên mạng nơ-rôn sâu giúp máy tính hiểu và tạo ra văn bản mạch lạc theo ngữ cảnh

NLP là lĩnh vực, còn LLM là một loại mô hình/công nghệ hiện đại được dùng để giải quyết các bài toán NLP.

# RAG (Retrieval-Augmented Generation - Sinh tăng cường tra cứu)
Khái niệm & Bản chất:
- Là một mô hình thiết kế (design pattern) trong đó quá trình sinh văn bản của LLM được bổ sung (augmented) bằng các thông tin tra cứu (retrieved) từ cơ sở tri thức bên ngoài (thường lưu dưới dạng Vector Store) ngay tại thời điểm truy vấn

- Giúp gắn kết câu trả lời của mô hình vào dữ liệu thực tế tin cậy (Grounding), hạn chế tình trạng mô hình bốc phét/ảo giác (Hallucination)

Nó gồm hai ý quan trọng:
Retrieval → tìm/truy xuất thông tin liên quan.
Generation → LLM dùng thông tin tìm được để tạo câu trả lời.
RAG = “Cho LLM tra tài liệu trước khi trả lời.”
Điểm đặc biệt là RAG không nhất thiết làm LLM thông minh hơn hay thay đổi kiến thức bên trong model. Nó chủ yếu cung cấp thêm context phù hợp tại thời điểm trả lời, giúp câu trả lời dựa trên dữ liệu bên ngoài, riêng tư hoặc mới hơn.

Nhầm lẫn RAG với Fine-tuning: Fine-tuning là thay đổi trọng số mô hình để học hành vi/văn phong, còn RAG là cung cấp tri thức động/mới nhất mà không thay đổi trọng số mô hình


# Agents (AI Agents)
Agent = một hệ thống AI có thể nhận mục tiêu → tự quyết định cần làm gì → sử dụng công cụ → quan sát kết quả → tiếp tục hành động để hoàn thành mục tiêu.

Nếu LLM giống “bộ não biết suy nghĩ và trả lời”, thì Agent giống “một nhân viên AI có bộ não đó và còn biết hành động”.

## !REMEMBER
- LLM = biết suy nghĩ/tạo câu trả lời
- RAG = cho LLM tra thêm tài liệu
- Agent = cho LLM khả năng quyết định và hành động bằng công cụ
- Multi-Agent = nhiều Agent phối hợp/phân chia công việc với nhau

Ví dụ đời thường: LLM = nhân viên tư vấn, RAG = nhân viên được phép tra kho tài liệu, còn Agent = nhân viên được giao mục tiêu, có công cụ và có thể tự thực hiện nhiều bước để hoàn thành việc.

# Evaluation Mindset
Evaluation-Driven Development (Phát triển dựa trên đánh giá): Định nghĩa tiêu chí và xây dựng hệ thống đánh giá (evaluation pipeline) trước khi bắt tay vào phát triển ứng dụng AI. 

 Vì đầu ra của LLM là dạng mở (open-ended), kiểm thử truyền thống (assert đúng/sai tuyệt đối) không còn áp dụng được
 Không chỉ quan tâm “model chạy được không?”, mà luôn nghĩ “làm sao biết nó chạy tốt, tốt đến mức nào, và thất bại ở đâu?”

 Evaluation Mindset = luôn đặt câu hỏi: “Làm sao tôi chứng minh hệ thống AI này thực sự tốt?”

#  Evaluation Mindset
