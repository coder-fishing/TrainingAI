# Tổng quan về Dữ liệu và Học máy (Machine Learning)

1. Các Khái niệm Cốt lõi về Dữ liệu

Data (Dữ liệu): Thông tin được tổ chức dưới dạng bảng, trong đó mỗi hàng đại diện cho một điểm dữ liệu (data point) riêng biệt.

Features (Đặc trưng): Các cột trong bảng dùng để mô tả thuộc tính hoặc đặc điểm của dữ liệu (ví dụ: tên, cân nặng, kích thước, hoặc màu sắc pixel của ảnh).

Labels (Nhãn): Một dạng đặc trưng đặc biệt mà chúng ta muốn mô hình dự đoán dựa trên các features khác (ví dụ: loại thú cưng, tình trạng bệnh, giá nhà).

2. Dữ liệu Có nhãn và Không nhãn

Labeled Data (Dữ liệu có nhãn): Dữ liệu đi kèm với cột nhãn rõ ràng (ví dụ: email có nhãn spam/ham, ảnh có nhãn trọng lượng hoặc loài vật).

Unlabeled Data (Dữ liệu không nhãn): Dữ liệu hoàn toàn không có cột nhãn đích cần dự đoán.

3. Học có giám sát (Supervised Learning)

Supervised learning là nhánh học máy sử dụng dữ liệu có nhãn để huấn luyện mô hình dự đoán nhãn cho dữ liệu mới. Mô hình tuân theo khung 3 bước:

Remember (Nhớ): Ghi nhớ tập dữ liệu huấn luyện.

Formulate (Định hình): Xây dựng quy luật hoặc mô hình từ dữ liệu.

Predict (Dự đoán): Dự đoán nhãn cho dữ liệu mới chưa biết.

4. Hai Loại Mô hình Supervised Learning Chính

Dựa vào bản chất của nhãn, học có giám sát chia thành hai loại chính:

A. Regression Models (Mô hình Hồi quy)
- Định nghĩa: Dự đoán dữ liệu dạng số (numerical data).
- Ví dụ: Dự đoán giá nhà, giá cổ phiếu, tuổi thọ bệnh nhân, hoặc thời gian xem video.
- Thuật toán phổ biến: Linear Regression, Decision Tree Regression, Random Forests, XGBoost.

B. Classification Models (Mô hình Phân loại)
- Định nghĩa: Dự đoán dữ liệu dạng danh mục/trạng thái (categorical data) thuộc một tập hợp hữu hạn các trạng thái (ví dụ: có/không, chó/mèo, spam/ham).
- Ví dụ: Phân loại email spam, nhận diện chữ số viết tay (MNIST), nhận diện vật thể ảnh (CIFAR-10), phân tích cảm xúc đánh giá phim.

5. Học không giám sát (Unsupervised Learning)

Unsupervised learning là nhánh học máy làm việc với dữ liệu không có nhãn (unlabeled data). Mục tiêu là khai thác cấu trúc ẩn, tìm ra điểm tương đồng và trích xuất tối đa thông tin từ dữ liệu mà không có nhãn đích hướng dẫn trước.

6. Các Nhánh Chính của Unsupervised Learning

A. Clustering Algorithms (Thuật toán phân cụm)
- Định nghĩa: Nhóm dữ liệu thành các cụm (clusters) dựa trên mức độ tương đồng giữa các phần tử.
- Ứng dụng: Phân khúc khách hàng, phân nhóm email, y học, phân loại gen.
- Các thuật toán phổ biến: K-means clustering, Hierarchical clustering, Density-based spatial clustering (DBSCAN), Gaussian mixture models.

B. Dimensionality Reduction Algorithms (Thuật toán giảm chiều dữ liệu) PCA
- Định nghĩa: Đơn giản hóa dữ liệu bằng cách giảm số lượng cột (đặc trưng/dimensions) mà vẫn giữ lại tối đa thông tin quan trọng.
- Ý nghĩa: Giúp tối ưu hóa việc lưu trữ, xử lý và trực quan hóa dữ liệu có không gian nhiều chiều phức tạp.

C. Generative Algorithms (Thuật toán tạo sinh)
- Định nghĩa: Các thuật toán có khả năng tự động tạo ra các điểm dữ liệu mới mang đặc điểm mô phỏng tương tự dữ liệu thực tế hiện có.

# Học Tăng Cường (reinforcement learning)
7. Học tăng cường (Reinforcement Learning)

- Định nghĩa: Là nhánh học máy mà ở đó **không có sẵn tập dữ liệu (dataset)** để huấn luyện theo cách thông thường, thay vào đó máy tính phải học cách thực hiện một nhiệm vụ thông qua việc tương tác với môi trường.
- Các thành phần chính:
  - **Agent (Tác nhân):** Đối tượng thực hiện hành động (ví dụ: robot).
  - **Environment (Môi trường):** Không gian mà agent di chuyển và tương tác.
  - **Rewards and Punishments (Phần thưởng và Hình phạt):** Hệ thống điểm số hoặc tín hiệu hướng dẫn agent đưa ra quyết định đúng đắn để đạt mục tiêu.
- Cơ chế hoạt động: Agent thực hiện thử-sai (trial and error), ghi nhận kết quả điểm số (phần thưởng khi đi đúng hướng, hình phạt khi gặp chướng ngại vật) để tự rút kinh nghiệm và tìm ra chiến lược tối ưu nhất.
- Ứng dụng thực tế:
  - **Trò chơi (Games):** Các chương trình chơi cờ vua, cờ vây (Go) hay game Atari (Super Mario, Breakout).
  - **Robot:** Giúp robot học cách nhặt hộp, dọn phòng hoặc điều khiển chuyển động.
  - **Xe tự hành (Self-driving cars):** Lên kế hoạch đường đi và xử lý tình huống trong các môi trường giao thông khác nhau.
