### kernel trick
- khi gặp case như BMI từ 18.5 tới 32 thì bị đi nvqs
  - thì dữ liệu bị cụt đi không có tuyến tính nữa
  - thì bây giờ dùng 1 cái gọi là kernel trick (dùng 1 hàm để biến cái hiện tại lên không gian cao nhơ, rồi tách ra)
    -   
    - nó gọi là hàm Phi(x)=> giờ dùng dữ liệu của hàm mới ( dùng cái mới để đi tìm hyperplane)
    - sau khi có các điểm mới thì ta tính tích vô hướng của 1 điểm với các điểm còn lại 
- Kernel function: tính toán với khoong gian chiều lớn hơn bằng cái input ban đầu

Những công thức nên nhớ
1. Decision Boundary
$$
w^Tx+b=0
$$

2. Hai đường biên của SVM
$$
w^Tx+b=1,\qquad w^Tx+b=-1
$$

3. Chiều rộng Margin
$$
\text{Margin width}=\frac{2}{\|w\|}
$$

4. Mục tiêu của Hard-Margin SVM
$$
\min_{w,b}\frac12\|w\|^2
$$

với điều kiện:
$$
y_i(w^Tx_i+b)\geq1
$$

#### RBF là gì trong SVM?
RBF (Radial Basis Function) là một loại Kernel thường được sử dụng trong Support Vector Machine (SVM) để phân loại dữ liệu phi tuyến (Non-linear).
Hiểu đơn giản:
- Linear SVM: Tìm một đường thẳng phân chia dữ liệu.
- SVM với RBF Kernel: Có thể tạo ra đường ranh giới cong, thậm chí khép kín, để phân chia dữ liệu phức tạp.  
-   
2. Công thức của RBF Kernel
$$
K(x_i,x_j)=e^{-\gamma\|x_i-x_j\|^2}
$$

Trong đó:
- \(x_i,x_j\): Hai điểm dữ liệu.
- \(\|x_i-x_j\|^2\): Bình phương khoảng cách Euclidean giữa hai điểm.
- \(\gamma\) (gamma): Hyperparameter điều chỉnh mức độ ảnh hưởng theo khoảng cách.
- \(e\): Hằng số Euler, khoảng 2.718.  
-   
- 
  - với các gamma thì nó khác

from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler

model = make_pipeline(

    SVC(

        C=1.0,

model.fit(X_train, y_train)

#### Hinge Loss
Hinge Loss là hàm mất mát phổ biến trong SVM:
$$
\boxed{L=\max(0,1-y(w^Tx+b))}
$$

Trong đó \(y\in\{-1,+1\}\).

 



- L càng lớn thì hyperplane càng lỏ => càng sai L =0 là ngon nhất
