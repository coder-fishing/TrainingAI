Trong toán học của mô hình tuyến tính, công thức thường có dạng quen thuộc:$$\text{Prediction} = (\text{Weight} \times \text{Feature}) + \text{Bias}$$(hoặc viết gọn là: $y = wx + b$)
- Ý nghĩa trực quan của hai khái niệm này:
- Weights (Trọng số - $w$):Là hệ số nhân với mỗi đặc trưng (feature).Nó quyết định mức độ ảnh hưởng của đặc trưng đó đến kết quả dự đoán. Ví dụ: Nếu trọng số của số lượng phòng là 50, nghĩa là cứ thêm 1 phòng thì giá nhà tăng thêm 50 đơn vị.
- Bias (Độ lệch / $b$):Là hằng số không gắn liền với bất kỳ đặc trưng nào.Nó đại diện cho giá trị cơ sở (baseline) — tức là giá trị dự đoán khi tất cả các đặc trưng đều bằng 0, hoặc giúp mô hình dịch chuyển đường thẳng cho phù hợp với tổng thể dữ liệu.

Slope (Độ dốc - $m$) $\leftrightarrow$ Weight (Trọng số - $w$). 

- Trong hình học: Thể hiện độ dốc của đường thẳng (tỷ lệ tăng lên theo chiều dọc so với chiều ngang).
- Trong Machine Learning: Là trọng số của đặc trưng, cho biết khi giá trị đặc trưng ($x$) tăng thêm 1 đơn vị, giá trị dự đoán ($y$) sẽ thay đổi bao nhiêu
- Y-intercept (Giao điểm trục tung - $b$) $\leftrightarrow$ Bias (Độ lệch - $b$):Trong hình học: Là độ cao tại điểm đường thẳng cắt trục $y$.Trong Machine Learning: Là bias, biểu thị giá trị cơ sở của nhãn khi tất cả các đặc trưng đầu vào đều bằng 0.
- Linear Equation (Phương trình tuyến tính - $y = mx + b$):Là công thức cơ bản của đường thẳng, trong đó $x$ là đặc trưng (feature), $y$ là dự đoán (prediction), $m$ là trọng số (weight) và $b$ là độ lệch (bias).

- Nói chung là công thức y= mx+b trong đó m là hệ số góc còn b là cái giao với trục y

#### Vậy khác nhau của 2 công thức đó là gì thì 

- với y = mx + b thì là cái chung tổng quảt còn 
- y = wx + b thì w đó là w1.x1 + .... wnxn + b



## CHI TIẾT VỀ HỒI QUY TUYẾN TÍNH (LINEAR REGRESSION), MSE & GRADIENT DESCENT

### 1. Linear Regression là gì?
Mô hình đơn giản với một feature $x$:

$$\hat{y} = mx + b$$

Trong đó:
- $x$: feature đầu vào.
- $y$: giá trị thực tế (actual value).
- $\hat{y}$: giá trị dự đoán (prediction).
- $m$: slope / weight (hệ số góc).
- $b$: intercept / bias (hệ số chặn / cơ sở).

### 2. MSE — Mean Squared Error & Loss Function
Để đo lường mức độ sai lệch giữa thực tế $y$ và dự đoán $\hat{y}$, ta dùng **MSE**:

$$MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$$

- **Loss Function:** Hàm số đo lường mức độ dự đoán tệ của mô hình. 
- **Mục tiêu training:** Tìm parameters ($m, b$) sao cho MSE / Loss đạt giá trị nhỏ nhất.

### 3. Gradient & Gradient Descent
* **Gradient:** Độ dốc của Loss function, cho biết hướng và mức độ thay đổi của Loss khi điều chỉnh parameters:
  $$\frac{\partial MSE}{\partial m}, \quad \frac{\partial MSE}{\partial b}$$
* **Gradient Descent:** Thuật toán tối ưu đi **ngược hướng Gradient** để tìm về điểm Loss thấp nhất.
* **Learning Rate ($\alpha$):** Kích thước bước đi của mỗi lần cập nhật.
  - Quá nhỏ $\rightarrow$ Chạy rất chậm, mất thời gian.
  - Quá lớn $\rightarrow$ Nhảy quá đà, dao động, không hội tụ.

### 4. Công thức cập nhật Parameters

$$m_{\text{new}} = m_{\text{old}} - \alpha \frac{\partial MSE}{\partial m}$$

$$b_{\text{new}} = b_{\text{old}} - \alpha \frac{\partial MSE}{\partial b}$$

### 5. Epoch & Least Squares
* **Epoch:** Một vòng mô hình duyệt qua **toàn bộ** tập dữ liệu training.
* **Least Squares (Bình phương tối thiểu):** Phương pháp toán học giúp tìm nghiệm tối ưu trực tiếp cho Linear Regression mà **không bắt buộc** phải dùng đến Gradient Descent hay Learning Rate.

## Hồi quy đa thức Polynomial Regression



````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# =====================
# 1. Data
# =====================

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 5, 10, 17, 26])

# =====================
# 2. Create model
# =====================

model = LinearRegression()

# =====================
# 3. Train
# =====================

model.fit(X, y)

# =====================
# 4. Predict
# =====================

y_pred = model.predict(X)

# =====================
# 5. Parameters
# =====================

print("Weight:", model.coef_)
print("Bias:", model.intercept_)

# =====================
# 6. Results
# =====================

print("Actual:", y)
print("Predicted:", y_pred)

# =====================
# 7. Visualize
# =====================

plt.scatter(X, y, label="Actual Data")
plt.plot(X, y_pred, label="Linear Regression")

plt.xlabel("X")
plt.ylabel("y")
plt.title("Linear Regression")
plt.legend()

plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# =====================
# 1. Data
# =====================

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 5, 10, 17, 26])

# =====================
# 2. Create Polynomial Features
# =====================

poly = PolynomialFeatures(
    degree=2,
    include_bias=False
)

X_poly = poly.fit_transform(X)

print("Original X:")
print(X)

print("\nPolynomial X:")
print(X_poly)

# =====================
# 3. Create model
# =====================

model = LinearRegression()

# =====================
# 4. Train
# =====================

model.fit(X_poly, y)

# =====================
# 5. Predict
# =====================

y_pred = model.predict(X_poly)

# =====================
# 6. Result
# =====================

print("\nWeights:", model.coef_)
print("Bias:", model.intercept_)
print("Prediction:", y_pred)

# =====================
# 7. Visualization
# =====================

X_curve = np.linspace(1, 5, 100).reshape(-1, 1)

X_curve_poly = poly.transform(X_curve)

y_curve = model.predict(X_curve_poly)

plt.scatter(X, y, label="Actual Data")
plt.plot(X_curve, y_curve, label="Polynomial Regression")

plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
````

degree (d) là Hyperparameter. Ta chọn nó trước training và dùng validation/cross-validation để tìm giá trị cho khả năng dự đoán dữ liệu mới tốt nhất, thay vì chọn degree chỉ vì nó fit training data đẹp nhất.

- Error = y - ŷ → sai bao nhiêu.
- Loss function → biến error thành một giá trị để model tối ưu.
- MAE/MSE → hai cách phổ biến để tổng hợp error thành loss/metric trong regression.



- MAE: "Trung bình tôi sai bao nhiêu?"
- MSE: "Trung bình tôi sai bao nhiêu, nhưng hãy phạt thật mạnh những lần sai lớn."

- loss là giá trị phạt
