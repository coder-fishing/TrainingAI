## Foundation models (mô hình nền tảng)
- Định nghĩa & Tiến hóa: Trình bày quá trình chuyển đổi từ các mô hình NLP truyền thống (chỉ làm 1 tác vụ như dịch thuật hay phân loại) $\rightarrow$ Large Language Models (LLMs) $\rightarrow$ Foundation Models (Mô hình nền tảng đa thức - multimodal, xử lý cả text, code, audio, image, video).
- Bản chất của AI Engineering: Sự xuất hiện của Foundation Models đã biến AI từ một lĩnh vực hàn lâm/nghiên cứu phức tạp thành một công cụ phát triển phần mềm mà bất kỳ lập trình viên nào cũng có thể truy cập và sử dụng thông qua API hoặc các mô hình mã nguồn mở.

## 2. Model Capabilities and Limitations (Khả năng & Giới hạn của mô hình)

Ở Chương 1, tác giả xem xét khả năng và giới hạn dưới góc độ **Lập kế hoạch ứng dụng (Planning AI Applications)**:

### Khả năng (Capabilities)
Phân tích 8 nhóm use-case thành công nổi bật nhất hiện nay:
* **Lập trình (Coding):** Hỗ trợ viết code, sửa lỗi và tự động hóa công việc lập trình.
* **Sản xuất hình ảnh/video (Image & Video Production):** Tạo và chỉnh sửa nội dung đa phương tiện.
* **Viết lách & Biên tập (Writing):** Soạn thảo, tóm tắt và chỉnh sửa văn bản.
* **Giáo dục & Trợ giảng (Education):** Cá nhân hóa việc học, tạo bài tập và tài liệu giảng dạy.
* **Chatbot hội thoại (Conversational Bots):** Xây dựng trợ lý ảo và chăm sóc khách hàng.
* **Tổng hợp thông tin (Information Aggregation):** Trích xuất và tóm tắt dữ liệu từ nhiều nguồn khác nhau.
* **Tổ chức dữ liệu phi cấu trúc (Data Organization):** Phân loại, dán nhãn và cấu trúc hóa dữ liệu.
* **Tự động hóa quy trình (Workflow Automation):** Tích hợp AI vào chuỗi công việc tự động.

### Giới hạn (Limitations) & Quản lý kỳ vọng (Setting Expectations)
* **Tính bất định:** Mô hình nền tảng mang bản chất xác suất (*probabilistic / non-deterministic*).
* **Độ chính xác:** Cần chấp nhận rằng AI không bao giờ đạt độ chính xác $100\%$ ngay từ đầu.
* **Quản lý rủi ro:** Việc nhận biết rõ giới hạn giúp đội ngũ đặt ra kỳ vọng đúng đắn với khách hàng và chuẩn bị các phương án dự phòng (*human-in-the-loop*, bảo trì liên tục).
