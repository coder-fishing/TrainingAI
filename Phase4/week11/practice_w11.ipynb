# AI Use Case Selection Matrix

| Tiêu chí | Use Case 1: Tự động hóa tạo Unit Test & Test Cases | Use Case 2: Phân tích & Trích xuất Yêu cầu Phần mềm (SRS) | Use Case 3: Trợ lý Chatbot Hỗ trợ Onboarding Lập trình viên mới |
| :--- | :--- | :--- | :--- |
| **User Problem** *(Vấn đề người dùng)* | Lập trình viên tiêu tốn từ 20–30% thời gian sprint để viết Unit Test thủ công, dẫn đến thiếu thời gian tối ưu logic chính và giảm test coverage. | Business Analyst (BA) mất nhiều ngày để đọc các tài liệu yêu cầu nghiệp vụ thô, cồng kềnh, dễ bỏ sót điều kiện biên và quy tắc logic. | Lập trình viên mới gia nhập team mất từ 2–3 tuần chỉ để đọc tài liệu hệ thống, hỏi đồng nghiệp cách cài đặt môi trường và tìm hiểu luồng code. |
| **AI Suitability** *(AI có phù hợp không?)* | **Rất phù hợp:** AI hiểu rõ cú pháp, logic mã nguồn và cấu trúc hàm để sinh ra các test case bao phủ cả trường hợp bình thường lẫn edge case. | **Phù hợp:** AI giỏi xử lý, tóm tắt dữ liệu phi cấu trúc và chuyển đổi văn bản tự nhiên thành danh sách yêu cầu dạng cấu trúc (User Stories/Acceptance Criteria). | **Rất phù hợp:** AI có khả năng đóng vai trợ lý hội thoại, truy vấn thông tin linh hoạt từ kho tài liệu nội bộ dựa trên câu hỏi tự nhiên. |
| **Expected Value** *(Giá trị kỳ vọng)* | - Rút ngắn 50% thời gian viết Unit Test.<br>- Tăng độ bao phủ mã nguồn (Test Coverage) lên trên 80%.<br>- Phát hiện sớm lỗi logic ngay ở mức hàm. | - Rút ngắn 60% thời gian phân tích tài liệu SRS ban đầu.<br>- Hạn chế sai sót/thiếu sót trong quy trình nghiệp vụ.<br>- Chuẩn hóa định dạng User Story cho toàn đội ngũ. | - Giảm thời gian onboarding lập trình viên mới từ 3 tuần xuống còn 1 tuần.<br>- Giảm 70% thời gian các Senior Dev phải bỏ ra để trả lời các câu hỏi lặp đi lặp lại. |
| **Risks** *(Rủi ro)* | - **Hallucination:** AI tự bịa ra tên hàm hoặc lớp test không tồn tại.<br>- **Security:** Lộ mã nguồn bảo mật nếu dùng API mô hình công cộng không có cam kết bảo vệ dữ liệu. | - Trích xuất thiếu hoặc hiểu sai quy tắc nghiệp vụ quan trọng.<br>- Không nhận diện được các mâu thuẫn logic trong tài liệu nguồn. | - Chatbot cung cấp hướng dẫn cài đặt hoặc thông tin kiến trúc cũ/đã deprecated do tài liệu nội bộ chưa cập nhật. |
| **Complexity** *(Độ phức tạp)* | **Low – Medium** | **Medium** | **High** |
| **Recommended Approach** *(Giải pháp đề xuất)* | Sử dụng mô hình mã nguồn mở chuyên về code (CodeLlama, Qwen-Coder) hoặc tích hợp Copilot Extension trực tiếp vào IDE, kèm prompt ràng buộc khung test (JUnit, Jest, PyTest). | Kỹ thuật **Prompt Engineering** nâng cao (Chain-of-Thought) kết hợp với **Structured Output (JSON/Markdown)** để bắt buộc AI xuất ra đúng định dạng User Story mong muốn. | Kiến trúc **RAG (Retrieval-Augmented Generation)** kết hợp Vector Database để truy xuất thông tin chính xác từ Wiki/Confluence nội bộ + Guardrails kiểm soát câu trả lời. |
