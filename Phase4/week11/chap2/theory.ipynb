# Chương 2: Understanding Foundation Models (Thấu hiểu các Mô hình Nền tảng)

Chương 2 trong cuốn sách *AI Engineering: Building Applications with Foundation Models* của Chip Huyen tập trung giải thích chuyên sâu về nguyên lý hoạt động, cấu trúc bên trong, quy trình huấn luyện và cơ chế sinh ra kết quả của các mô hình nền tảng (Foundation Models).

---

## 1. Dữ liệu huấn luyện (Training Data)

Khả năng và giới hạn của mô hình phụ thuộc trực tiếp vào tập dữ liệu tiền huấn luyện (*Pre-training Data*).

* **Phân bố dữ liệu (Data Distribution):** Mô hình chỉ hoạt động tốt trên các tác vụ, miền tri thức hoặc ngôn ngữ có mật độ xuất hiện đủ lớn trong tập dữ liệu huấn luyện.
* **Mô hình đa ngôn ngữ (Multilingual Models):** Thách thức lớn về sự mất cân bằng dữ liệu ngôn ngữ (tiếng Anh và một số ngôn ngữ phổ biến chiếm tỷ trọng vượt trội so với các ngôn ngữ ít tài nguyên).
* **Mô hình chuyên biệt theo miền (Domain-Specific Models):** Việc xây dựng các mô hình chuyên biệt (cho y tế, pháp lý, tài chính hoặc lập trình) giúp tối ưu hóa hiệu năng và giảm kích thước mô hình so với mô hình tổng quát.

---

## 2. Mô hình hóa & Kiến trúc (Modeling & Architecture)

* **Kiến trúc Transformer:** 
  * Giải thích lý do vì sao kiến trúc Transformer (đặc biệt là cơ chế Self-Attention) trở thành chuẩn mực cho xử lý ngôn ngữ tự nhiên và mô hình đa thức (multimodal).
  * Hạn chế của Transformer về độ phức tạp tính toán theo độ dài chuỗi đầu vào ($O(N^2)$) và tiềm năng của các kiến trúc thay thế (như Mamba/SSM).
* **Kích thước mô hình (Model Size):** 
  * 3 chỉ số định lượng quy mô chính: **Số lượng tham số** (*Parameters*), **Số lượng token huấn luyện** (*Training Tokens*), và **Chi phí tính toán** (*FLOPs*).
* **Quy luật mở rộng (Scaling Laws):** 
  * Mối quan hệ định lượng giữa ngân sách tính toán, kích thước mô hình và lượng dữ liệu để đạt hiệu năng tối ưu (ví dụ: định luật Chinchilla Scaling Law).
  * Các điểm nghẽn về hạ tầng tính toán, giới hạn năng lượng và việc cạn kệt dữ liệu chất lượng cao (*Scaling Bottlenecks*).

---

## 3. Tinh chỉnh sau huấn luyện (Post-Training & Alignment)

Mô hình sau giai đoạn Pre-training thực chất chỉ là một hệ thống hoàn thành chuỗi văn bản (text completion). Giai đoạn Post-training biến nó thành trợ lý hữu ích và an toàn.

* **Supervised Finetuning (SFT):** Tinh chỉnh có giám sát trên các cặp dữ liệu (*Instruction - Response*) để huấn luyện mô hình biết cách tuân thủ chỉ dẫn, trả lời câu hỏi và hội thoại.
* **Preference Finetuning (Căn chỉnh theo sở thích):**
  * **RLHF (Reinforcement Learning from Human Feedback):** Sử dụng mô hình thưởng (Reward Model) và thuật toán PPO để điều hướng câu trả lời của AI theo các tiêu chí: Hữu ích (*Helpful*), Trung thực (*Honest*), và Không độc hại (*Harmless*).
  * **DPO (Direct Preference Optimization):** Phương pháp tối ưu hóa trực tiếp dựa trên dữ liệu so sánh cặp (preferred vs. dispreferred) giúp đơn giản hóa quy trình huấn luyện không cần qua Reward Model riêng biệt.

---

## 4. Cơ chế lấy mẫu & Sinh kết quả (Sampling & Generation)

* **Nguyên lý cơ bản:** Mô hình không tạo ra toàn bộ câu cùng lúc mà sinh từng token một dựa trên phân bố xác suất của token tiếp theo ($P(w_t | w_1, w_2, ..., w_{t-1})$).
* **Chiến lược lấy mẫu (Sampling Strategies):**
  * **Temperature:** Điều chỉnh độ phẳng của phân bố xác suất. Temperature thấp làm phản hồi nhất quán/tiêu chuẩn; Temperature cao làm phản hồi đa dạng/sáng tạo hơn.
  * **Top-k & Top-p (Nucleus Sampling):** Giới hạn không gian lựa chọn token trong top $k$ token có xác suất cao nhất hoặc tập hợp token có tổng xác suất tích lũy đạt ngưỡng $p$.
* **Test-Time Compute:** Kỹ thuật tăng cường năng lực suy luận bằng cách dành nhiều tài nguyên tính toán hơn tại thời điểm sinh phản hồi (như Chain-of-Thought, Best-of-N sampling).
* **Đầu ra có cấu trúc (Structured Outputs):** Kỹ thuật áp đặt ràng buộc ngữ pháp (Grammar-constrained decoding) để bắt buộc mô hình trả về định dạng chuẩn xác (JSON, XML, SQL).

---

## 5. Bản chất xác suất & Hiện tượng Ảo giác (Probabilistic Nature & Hallucination)

* **Bản chất xác suất:** Mô hình là một hệ thống không định hình (*non-deterministic*). Nó tính toán khả năng xuất hiện của chuỗi từ chứ không truy vấn trực tiếp từ một cơ sở dữ liệu sự thật khách quan.
* **Nguyên nhân gốc rễ của Ảo giác (Hallucination):**
  * **Từ Pre-training:** Mô hình bị ép buộc phải sinh câu trả lời ngay cả khi không có đủ thông tin trong dữ liệu huấn luyện (dẫn đến việc "bịa" thông tin có vẻ hợp lý).
  * **Từ Sampling:** Việc sử dụng các chiến lược sampling tạo ra tính ngẫu nhiên làm tăng rủi ro lệch khỏi sự thật.
  * **Mâu thuẫn căn chỉnh:** Tinh chỉnh quá mức (Over-alignment) có thể khiến mô hình tự tin đưa ra câu trả lời sai thay vì từ chối trả lời.
