# Evaluate AI Systems (Đánh giá Hệ thống AI)

### 1. Triết lý & Khung Tư duy Đánh giá (Evaluation Mindset & Framework)

* **Evaluation-Driven Development (EDD) - Phát triển dựa trên Đánh giá** - Phương pháp luận yêu cầu kỹ sư phải thiết lập tiêu chí đánh giá và tập dữ liệu kiểm thử (Eval Dataset) trước khi viết prompt phức tạp hoặc chọn mô hình. Ví dụ: Chuẩn bị 100 câu hỏi mẫu cùng đáp án chuẩn trước khi bắt tay vào thiết kế prompt chatbot.
* **Continuous Evaluation - Đánh giá liên tục** - Quá trình kiểm thử không dừng lại khi ra mắt sản phẩm (Launch) mà phải được vận hành liên tục trên môi trường Production để theo dõi dữ liệu thực tế của người dùng theo thời gian.
* **Component-Level / Modular Evaluation - Đánh giá theo từng mắt xích** - Kiểm tra độc lập từng thành phần cấu thành hệ thống (Retriever, Re-ranker, Prompt Generator, LLM) thay vì chỉ chấm điểm đầu ra cuối cùng. Ví dụ: Trong hệ thống RAG, đánh giá riêng xem khâu tìm kiếm (Retriever) lấy đúng văn bản hay không, và khâu tổng hợp (LLM) có viết đúng sự thật từ văn bản đó không.
* **End-to-End Evaluation - Đánh giá toàn chuỗi** - Chấm điểm chất lượng của câu trả lời cuối cùng mà người dùng nhận được mà không phân tích sâu vào các bước xử lý trung gian.

---

### 2. Các Tiêu chí Đánh giá Ứng dụng AI (Evaluation Criteria)

* **Domain-Specific Capability - Năng lực chuyên môn ngành** - Thước đo mức độ am hiểu và xử lý chính xác các tác vụ trong lĩnh vực chuyên sâu như Y tế, Pháp lý, Tài chính, Lập trình hay Toán học.
* **Factual Consistency (Hallucination Detection) - Độ trung thực thông tin** - Kiểm tra xem câu trả lời của mô hình có hoàn toàn dựa trên dữ liệu được cấp hay tự ý bịa đặt thông tin. Ví dụ: Kiểm tra xem thông tin giá sản phẩm mà chatbot đưa ra có đúng với tài liệu quy định hay không.
* **Instruction-Following Capability - Khả năng tuân thủ chỉ dẫn** - Đánh giá việc mô hình tuân thủ các ràng buộc về định dạng đầu ra (JSON, XML, Markdown), giới hạn số từ, hoặc ngôn ngữ phản hồi theo đúng prompt yêu cầu.
* **Cost & Latency Trade-offs - Đánh đổi Chi phí & Độ trễ** - Cân đối giữa khả năng của mô hình với ngân sách tài chính (Token Cost) và trải nghiệm người dùng (Latency).

---

### 3. Phương pháp Đánh giá & Thứ tự Ưu tiên (Evaluation Methods & Priority)

* **Rule-Based / Exact Match Evaluation (Ưu tiên 1) - Đánh giá dựa trên quy tắc** - Sử dụng code tự động kiểm tra định dạng cứng (Code Syntax, JSON Schema, Regex, Unit Test). Chạy cực nhanh, chi phí bằng 0 và đạt độ chính xác 100%. Ví dụ: Chạy pytest để kiểm tra đoạn code Python do AI sinh ra.
* **AI-as-a-Judge (Ưu tiên 2) - Dùng AI làm Giám khảo** - Sử dụng một mô hình AI có năng lực cao (như GPT-4o, Claude 3.5 Sonnet) để tự động chấm điểm các phản hồi dạng văn bản tự do (Open-ended output) dựa trên bộ tiêu chí chấm điểm (Rubrics) có sẵn.
* **Human-in-the-Loop Evaluation (Ưu tiên 3) - Con người thẩm định** - Sử dụng chuyên gia hoặc người dùng thật để chấm điểm. Thường dùng để tạo tập dữ liệu mẫu ban đầu (Gold Standard) hoặc kiểm tra ngẫu nhiên (spot-check) dữ liệu trên Production.
* **Pairwise Comparison - So sánh cặp** - Phương pháp cho AI Judge hoặc con người đọc hai câu trả lời từ hai mô hình khác nhau (A và B) cho cùng một câu hỏi và chọn ra câu trả lời tốt hơn.

---

### 4. Lựa chọn Mô hình & Hạ tầng (Model Selection & Deployment Options)

* **Proprietary Models (Closed-source) - Mô hình đóng** - Mô hình độc quyền dùng qua API (OpenAI, Anthropic). Phù hợp cho giai đoạn thử nghiệm (PoC), cần triển khai nhanh và chưa có nhân sự hạ tầng.
* **Open-Source Models - Mô hình mã nguồn mở tự host** - Mô hình cho phép tải về tự vận hành (Llama 3, Qwen). Phù hợp cho ứng dụng quy mô lớn (tối ưu chi phí) hoặc cần bảo mật dữ liệu tuyệt đối tại chỗ (On-premise).
* **Public Benchmarks & Leaderboards - Bảng xếp hạng công khai** - Các bộ đề test chuẩn quốc tế (MMLU, HumanEval, Chatbot Arena). Chỉ mang tính chất tham khảo ban đầu vì dễ bị lệch so với dữ liệu thực tế của doanh nghiệp.
* **Time-to-First-Token (TTFT) - Thời gian phản hồi chữ đầu tiên** - Khoảng thời gian tính từ lúc người dùng gửi câu hỏi đến khi từ (token) đầu tiên hiện lên màn hình.
* **Inter-Token Latency (ITL) / Throughput - Tốc độ nhả chữ** - Số lượng token mà mô hình có thể tạo ra trong 1 giây (đơn vị: tokens/second).

### 5. Các Bẫy & Lỗi Hiểu Lầm Cần Tránh (Common Misconceptions & Biases)

* **Data Contamination - Lộ đề test (Trôi dữ liệu)** - Hiện tượng câu hỏi trong bộ benchmark công khai đã bị lọt vào tập dữ liệu huấn luyện của mô hình. Dẫn đến việc mô hình học thuộc lòng bài test, điểm số cao vút trên mạng nhưng ra sản phẩm thật lại rất kém.
* **Data Drift / Model Drift - Lạc hậu dữ liệu** - Xu hướng chất lượng mô hình suy giảm theo thời gian do thói quen, từ ngữ hoặc bối cảnh của người dùng thực tế đã thay đổi so với lúc thiết kế ban đầu.
* **Position Bias - Thiên vị vị trí** - Lỗi của AI Judge khi có xu hướng ưu tiên chọn câu trả lời xuất hiện ở vị trí đầu tiên (hoặc vị trí thứ hai) trong bài so sánh cặp.
* **Verbosity Bias - Thiên vị độ dài** - Lỗi của AI Judge khi có xu hướng chấm điểm cao hơn cho các câu trả lời dài dòng, trình bày đẹp mắt dù nội dung có thể chứa thông tin thừa hoặc sai nhẹ.
* **Self-Enhancement Bias - Thiên vị bản thân** - Lỗi của AI Judge khi có xu hướng thiên vị và chấm điểm cao hơn cho các câu trả lời do chính mô hình đó (hoặc cùng họ mô hình) tạo ra.

---

### 6. Quy trình Xây dựng Pipeline Đánh giá Thực tế (Code & Execution)

* **Bước 1: Chuẩn bị Eval Dataset (Gold Standard)** - Xây dựng tập hợp 50-500 câu hỏi mẫu đại diện cho người dùng thật, kèm theo đáp án chuẩn hoặc quy tắc kiểm tra.
* **Bước 2: Tự động hóa kiểm thử bằng Code** - Tích hợp các thư viện đánh giá (như deepeval, ragas hoặc viết script custom) vào quy trình CI/CD của dự án.
* **Bước 3: Định kỳ rà soát & Cập nhật** - Liên tục bổ sung các trường hợp AI trả lời sai ngoài thực tế (Edge cases) vào tập Eval Dataset để bài test ngày càng hoàn thiện.
