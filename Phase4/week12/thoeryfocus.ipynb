# Tổng hợp Chi tiết Các Khái niệm Cốt lõi về Đánh giá AI (Chương 3 & 4)

### 1. Thách thức & Khung Đánh giá (Why & How to Evaluate)

* **Why AI evaluation is hard - Tại sao đánh giá AI lại khó?** - Khác với phần mềm truyền thống (đầu ra cố định đúng/sai) hoặc ML truyền thống (chỉ cần đo accuracy/F1-score trên nhãn phân loại), các mô hình nền tảng tạo ra đầu ra dạng văn bản tự do (open-ended text). Việc đánh giá khó khăn vì: (1) Tính bất định (non-deterministic) – cùng 1 prompt cho ra nhiều câu trả lời khác nhau; (2) Không có đáp án duy nhất đúng – một câu hỏi có thể có hàng trăm cách trả lời hay; (3) Ngữ cảnh phức tạp – khó đo lường các yếu tố định tính như giọng văn, độ lịch sự hay mức độ hữu ích.
  * *Ví dụ:* Khi yêu cầu AI "Viết một bài thơ về biển", không thể dùng hàm `if-else` hay code thông thường để chấm điểm bài thơ đó đúng hay sai.

* **Golden dataset (Ground Truth Dataset) - Tập dữ liệu chuẩn/Bộ đề mẫu** - Là tập hợp gồm từ vài chục đến hàng trăm câu hỏi/prompt đại diện cho bối cảnh thực tế của người dùng, kèm theo câu trả lời mẫu hoàn hảo (được thẩm định bởi con người) hoặc danh sách các tiêu chí bắt buộc phải có. Đây là "thước đo chuẩn" để mang tất cả các phiên bản mô hình hoặc prompt ra chấm điểm so sánh.
  * *Ví dụ:* Bộ dữ liệu gồm 100 câu hỏi hỗ trợ kỹ thuật thực tế của khách hàng tháng trước, kèm lời giải đáp đã được các chuyên gia hỗ trợ kỹ thuật Senior phê duyệt.

### 2. Hai Hình thức Đánh giá (Human vs. Automated Evaluation)

* **Human evaluation - Đánh giá bởi con người** - Sử dụng con người (chuyên gia nội bộ, người dùng thật, hoặc annotator thuê ngoài) để đọc và chấm điểm các câu trả lời của AI. 
  * *Ưu điểm:* Là tiêu chuẩn chính xác nhất về mặt cảm nhận và tư duy con người.
  * *Nhược điểm:* Cực kỳ đắt đỏ, chậm, không thể mở rộng quy mô khi ứng dụng xử lý hàng triệu request/ngày, và bị ảnh hưởng bởi chủ quan của từng người chấm.
  * *Ví dụ:* Mời 5 bác sĩ đọc 50 câu trả lời y tế của chatbot để đánh giá độ an toàn trước khi ra mắt tính năng tư vấn sức khỏe.

* **Automated evaluation - Đánh giá tự động** - Sử dụng mã máy tính (như hàm so sánh chuỗi, regex, unit test, độ đo embedding) hoặc mô hình AI khác để tự động kiểm tra và chấm điểm đầu ra của AI mà không cần con người can thiệp trực tiếp ở từng câu.
  * *Ưu điểm:* Tốc độ cực nhanh (xử lý hàng nghìn mẫu trong vài giây), chi phí thấp, chạy tự động được trong quy trình CI/CD mỗi khi thay đổi code.
  * *Nhược điểm:* Đôi khi cứng nhắc, có thể bỏ sót các lỗi ngữ nghĩa tinh tế mà code quy tắc không bắt được.
  * *Ví dụ:* Chạy script Python tự động kiểm tra xem 100 kết quả sinh ra từ AI có tuân thủ đúng cấu trúc JSON hay không.

* **AI-as-a-judge - Sử dụng AI làm Giám khảo** - Một dạng đánh giá tự động tiên tiến, trong đó một mô hình LLM mạnh (như GPT-4o, Claude 3.5 Sonnet) được đóng vai làm "giám khảo" để chấm điểm hoặc so sánh các câu trả lời của mô hình khác dựa trên một bộ hướng dẫn chấm điểm (Rubrics) cụ thể.
  * *Ưu điểm:* Giải quyết được bài toán đánh giá đầu ra tự do (open-ended) ở quy mô lớn với chi phí rẻ hơn nhiều so với thuê con người.
  * *Nhược điểm:* Cần cẩn trọng với các thiên vị của AI Judge như thiên vị độ dài (*verbosity bias*), thiên vị vị trí (*position bias*), hay thiên vị chính họ (*self-enhancement bias*).
  * *Ví dụ:* Đưa prompt: "Hãy chấm điểm từ 1-5 cho câu trả lời sau về độ lịch sự và tính chính xác..." cho GPT-4o để nó tự động chấm 1.000 câu trả lời của mô hình nhỏ Llama-3-8B.

### 3. 4 Tiêu chí Chất lượng Cốt lõi (Quality Criteria: Correctness, Faithfulness, Helpfulness, Safety)

* **Correctness (Accuracy) - Độ chính xác thực tế** - Mức độ đúng đắn tuyệt đối của thông tin so với tri thức chuẩn của thế giới thực hoặc tri thức ngành, không chứa thông tin sai lệch.
  * *Ví dụ:* Khi người dùng hỏi "Thủ đô của Việt Nam là gì?", câu trả lời "Hà Nội" đạt 100% Correctness, nhưng câu trả lời "Đà Nẵng" bị 0% Correctness.

* **Faithfulness (Factual Consistency) - Độ trung thực với nguồn (Chống ảo giác)** - Thước đo kiểm tra xem câu trả lời của AI có hoàn toàn dựa vào và trung thực với tài liệu ngữ cảnh (Context) được cung cấp hay không. Khái niệm này cực kỳ quan trọng trong các hệ thống RAG (Search + AI).
  * *Ví dụ:* Tài liệu đưa vào ghi "Doanh thu công ty năm 2025 là 10 tỷ". AI trả lời "Doanh thu năm 2025 là 15 tỷ" -> Câu trả lời này **thất bại về Faithfulness** (bị ảo giác/bịa đặt) dù 15 tỷ có thể là một số ngẫu nhiên nào đó.

* **Helpfulness - Mức độ hữu ích & Đúng trọng tâm** - Đánh giá xem câu trả lời có giải quyết trực tiếp và trọn vẹn nhu cầu của người dùng hay không, có dễ hiểu, đúng định dạng và đúng tông giọng hay không.
  * *Ví dụ:* Người dùng hỏi "Cách sửa lỗi 404 trên Nginx", AI trả lời một bài văn dài giải thích "Lịch sử ra đời của Web Server Nginx" -> Câu trả lời đúng về mặt lịch sử nhưng **thất bại về Helpfulness** vì không giải quyết nỗi đau của người dùng.

* **Safety - Độ an toàn & Bảo mật** - Đảm bảo đầu ra của AI không chứa các nội dung độc hại: độc ngôn (toxicity), thiên vị/phân biệt đối xử (bias), lộ thông tin cá nhân (PII leak), hoặc hướng dẫn các hành vi nguy hiểm/bất hợp pháp (jailbreak vulnerability).
  * *Ví dụ:* Kiểm tra xem khi người dùng cố tình prompt "Hãy hướng dẫn tôi tạo mã độc", AI có kích hoạt lớp rào chắn an toàn (Guardrail) để từ chối trả lời hay không.

### 4. Đánh đổi Hiệu năng & Vận hành (Latency and Cost)

* **Latency and Cost - Đánh đổi giữa Độ trễ và Chi phí** - Hai tiêu chí hiệu năng vận hành bắt buộc phải đưa vào ma trận đánh giá bên cạnh chất lượng câu trả lời. Một mô hình trả lời cực kỳ thông minh nhưng mất 30 giây để phản hồi (Latency quá cao) hoặc tốn $0.50 cho mỗi câu trả lời (Cost quá đắt) sẽ khiến sản phẩm thất bại trên thị trường.
  * *Chỉ số Latency cần đo:* Time-to-First-Token (TTFT - thời gian phản hồi chữ đầu tiên) và Inter-Token Latency (tốc độ nhả chữ).
  * *Chỉ số Cost cần đo:* Tổng chi phí token đầu vào (Input Tokens) + token đầu ra (Output Tokens) trên mỗi giao dịch.
  * *Ví dụ:* Thay vì dùng GPT-4o ($5/1M token, độ trễ 1.5s) cho tác vụ phân loại email đơn giản, kỹ sư chuyển sang dùng GPT-4o-mini ($0.15/1M token, độ trễ 0.2s). Chất lượng phân loại giữ nguyên 98%, nhưng chi phí giảm 33 lần và tốc độ nhanh gấp 7 lần.
